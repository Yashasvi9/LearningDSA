```python
def isPrime(n):
    prime = [True] * (n+1)
    p = 2
    while p*p <= n:
        if prime[p]:
            for i in range(p*p, n+1, p):
                prime[i] = False
        p += 1
    res = []
    for i in range(2, n+1):
        if prime[i]:
            res.append(i)
    return res
if __name__ == "__main__":
    n = 35
    res = isPrime(n)
    print(res)
```

```python
def extended_gcd(a, b):
    if a == 0:
        return b, 0, 1
    gcd, x1, y1 = extended_gcd(b%a, a)
    x = y1-(b//a)*x1
    y = x1
    return gcd, x, y

def modInverse(b, m):
    gcd, x, _ = extended_gcd(b, m)
    if gcd != 1:
        return -1
    return (x%m + m) % m

def modDivide(a, b, m):
    a = a%m
    inv = modInverse(b, m)
    if inv == -1:
        return -1
    return (a*inv)%m

if __name__ == "__main__":
    a = 10
    b = 2
    m = 13
    res = modDivide(a, b, m)
    print(res)
```

```python
def modExp(x, n, M):
    res = 1
    while n > 0:
        if n%2 == 1:
            res = (res*x) % M
            n -= 1
        else:
            x = (x*x)%M
        n //= 2
    return res

if __name__ == "__main__":
    x = 3
    n = 2
    m = 4
    print(modExp(x, n, m))
```
