```python
#Checking Prime Number
import math

def isPrime(n):
    if n <= 1:
        return False
    
    if n == 2 or n == 3:
        return True
    
    if n%2 == 0 or n%3==0:
        return False
    
    i = 5
    while i <= math.sqrt(n):
        if n%i == 0 or n%(i+2) == 0:
            return False
        
        i += 6
        
    return True

if __name__ == '__main__':
    n = 7
    if(isPrime(n)):
        print('true')
    else:
        print('false')
```

```python
#Sieve of Eratosthenes (All prime number upto a given number)

def sieve(n):
    prime = [True] * (n+1)
    p = 2
    
    while p*p <= n:
        if prime[p]:
            for i in range(p*p, n+1, p):
                prime[i] = False
        p += 1
        
    res = []
    for p in range(2,n+1):
        if prime[p]:
            res.append(p)
            
    return res

if __name__ == "__main__":
    n = 35
    res = sieve(n)
    for ele in res:
        print(ele, end = ' ')
```

```python
#Binary Exponentiation

def powMod(x, n, M):
    res = 1
    
    while n >= 1:
        if n%2 == 1:
            res = (res*x) % M

            n -= 1
        else:
            x = (x*x)%M
            n //= 2
    return res

if __name__ == '__main__':
    x, n, M =3, 2, 4
    print(powMod(x, n, M))
```

```python
#Approach 1

def decToBinary(n):
    binArr = []
    
    while n>0:
        bit = n%2
        binArr.append(str(bit))
        n //= 2
    
    binArr.reverse()
    return "".join(binArr)

if __name__ == "__main__":
    n = 12
    print(decToBinary(n))
```

```python
#Approach 2

def decToBinaryrec(n, binArr):
    if n == 0:
        return 
    decToBinaryrec(n//2, binArr)

    binArr.append(str(n%2))
    
def decToBinary(n):
    if n==0:
        return "0"
    binArr = []
    decToBinaryrec(n,binArr)
    return "".join(binArr)


if __name__ == '__main__':
    n = 12
    print(decToBinary(n))
```

```python
#Approach-1

def gcd(a,b):
    if a == 0 or b == 0:
        return max(a,b)
    
    result = min(a,b)
    
    while result > 0:
        if a%result == 0 and b%result == 0:
            break
        result -= 1
        
    return result

if __name__ == '__main__':
    a = 36
    b = 60
    print(gcd(a,b))
```

```python
#Approach-2

def gcd(a,b):
    if a == 0:
        return b
    if b == 0:
        return b
    
    if a==b:
        return a
    
    if a>b:
        return gcd(a - b,b)
    
    return gcd(a,b - a)

if __name__ == '__main__':
    a = 36
    b = 60
    print(gcd(a,b))
```

```python
#Approach-3    

def gcd(a, b):
    if a == 0:
        return b
    if b == 0:
        return a 
    
    if a == b:
        return a 
    
    if a>b:
        if a%b == 0:
            return b
        return gcd(a-b, b)
    
    if b>a:
        if b%a == 0:
            return a 
        return gcd(a, b-a)

if __name__ == '__main__':
    a = 36
    b = 60
    print(gcd(a,b))
```

```python
#Approach 1

def lcm(a,b):
    g = max(a,b)
    s = min(a,b)
    
    for i in range(g, a*b+1, g):
        if i%s == 0:
            return i
    return a*b

if __name__ == "__main__":
    a = 10
    b = 5
    print(lcm(a,b))
    
```

```python
#Approach - 2

def divBy13(s):
    length = len(s)
    if length == 1 and s[0] == '0':
        return True
    
    if length%3 == 1:
        s += "00"
        length += 2
    elif length%3 == 2:
        s += "0"
        length += 1
    
    sum_ = 0
    p = 1
    
    i = length-1
    
    while i >= 0:
        group = 0
        group += int(s[i])
        i -= 1
        group += int(s[i])*10
        i -= 1
        group += int(s[i])*100
        i -= 1
        
        sum_ += group*p
        p *= -1  #reverses the sign of each group after an iteration. For example in 1st iteration p = 1 then in next p becomes -1 and group changes to -group
    
    sum_ = abs(sum_)
    return sum_%13 == 0

if __name__ == '__main__':
    s = "2911285"
    isDivisible = divBy13(s)
    
    if isDivisible:
        print('true')
    else:
        print('false')
```

```python
#Binary Search Approach 

def floorSqrt(n):
    
    lo = 1
    hi = 1
    res = 1
    
    while lo <= hi:
        mid = lo + (hi-lo)//2
        
        if mid*mid <= n:
            res = mid
            lo = mid + 1
        
        else:
            hi = mid + 1
    return res

if __name__ == '__main__':
    n = 11
    print(floorSqrt(n))
```

```python
#Approach 1

def power(base, exp, mod):
    result = 1
    base %= mod
    while exp > 0:
        if exp%2 == 1:
            result = (result*base)%mod
        base = (base*base) % mod
        exp //= 2
    return result

def gcd(a,b):
    while b:
        a,b = b, a%b
    return a

def modInverse(b,m):
    return power(b,m-2,m)

def modDivide(a,b,m):
    if b == 0 or gcd(b,m) != 1:
        return -1
    inv = modInverse(b,m)
    return (a*inv)%m

if __name__ == '__main__':
    a, b, m = 10, 2, 13
    print(modDivide(a, b, m))
```

```python
#Apporach 2

def gcdExtended(a,b):
    if a == 0:
        return b, 0, 1
    gcd, x1, y1 = gcdExtended(b%a, a)
    x = y1-(b//a)*x1
    y = x1
    return gcd, x, y

def modInverse(b, M):
    gcd, x, _ = gcdExtended(b,M)
    if gcd != 1:
        return -1
    return(x % M + M)%M

def modDivide(a, b, M):
    a %= M
    inv = modInverse(b, M)
    if inv == -1:
        return -1
    return (a * inv) % M

if __name__ == '__main__':
    a = 10
    b = 2
    M = 13
    result = modDivide(a, b, M)
    print(result)
```

```python
#Approach 1

def nCr (n, r):
    if r > n:
        return 0
    if r == 0 or r == n:
        return 1

    return nCr(n-1, r-1) + nCr(n-1, r)

if __name__ == '__main__':
    n = 5
    r = 2
    print(nCr(n, r))
```

```python
#Apporach 2
def fact(n):
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

def nCr (n, r):
    if r > n:
        return 0
    if r == 0 or r == n:
        return 1

    return fact(n)//((fact(r))*(fact(n-r)))

if __name__ == '__main__':
    n = 5
    r = 2
    print(nCr(n, r))
```

```python
#Apporach 3

def Multiplier(start, end):
    if start == end:
        return start
    res = 1
    while start <= end:
        res *= start
        start += 1
    return res

def nCr(n, r):
    if n<r:
        return 0
    if r == 0 or r == n:
        return 1

    max_val = max(r,n-r)
    min_val = min(r, n-r)
    nume = Multiplier(max_val+1, n)
    deno = Multiplier(1, min_val)
    return nume//deno

if __name__ == '__main__':
    n = 5
    r = 2
    print(nCr(n, r))
```

```python
#Approach 2

def etf(n):

    result = n
    p = 2
    while p*p <= n:
        if n%p == 0:
            while n%p == 0:
                n //= p
            result -= result//p
        p += 1

    if n>1:
        result -= result //n
    return result

if __name__ == '__main__':
    n = 11
    print(etf(n))
```

```python
def sqrt(n):
    low = 1
    hi = n
    res = 1
    while low <= hi:
        mid = low +(hi-low)//2
        if mid*mid <= n:
            res = mid
            low = mid + 1
        else:
            hi = mid - 1
    return res
if __name__ == '__main__':
    n = 4
    print(sqrt(n))
```
